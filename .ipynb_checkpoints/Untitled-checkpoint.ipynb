```python
df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df[df['Review Rating'].isnull()]
```

```python
df.groupby('Category')['Review Rating'].agg(['count','median','mean'])
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df= df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
total_revenue = df['purchase_amount'].sum()
avg_purchase = df['purchase_amount'].mean()

print("Total Revenue:", total_revenue)
print("Average Purchase:", avg_purchase)
```

```python
category_sales = df.groupby('category')['purchase_amount'].sum().sort_values(ascending=False)

category_sales
```

```python
df.groupby('gender')['purchase_amount'].agg(['count','sum','mean'])
```

```python
df.groupby('subscription_status')['purchase_amount'].agg(['count','sum','mean'])
```

```python
labels = ['Young Adult', 'Adult', 'Midle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(15)
```

```python
df.groupby('age_group')['purchase_amount'].agg(['count','sum','mean'])
```

```python
frequency_mapping = {
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
df.groupby('discount_applied')['purchase_amount'].agg(['count','sum','mean'])
```

```python
df.groupby('season')['purchase_amount'].agg(['count','sum','mean'])
```
